```r
quiet_library <- function(...) { suppressPackageStartupMessages(library(...)) }
quiet_library(hise)
quiet_library(dplyr)
quiet_library(purrr)
```

## Read GSEA results stored in HISE

```r
hallmark_uuid <- list("4f05f2a0-59d2-4d04-b948-70d541613e71")
```

```r
hallmark_file <- cacheFiles(hallmark_uuid)
```

```r
no_psm_hallmark_uuid <- list("0b5ce7b5-ac2e-4704-9ab4-6d7864b1bb26")
```

```r
no_psm_hallmark_file <- cacheFiles(no_psm_hallmark_uuid)
```

```r
hallmark_gsea <- read.csv(hallmark_file)
no_psm_hallmark_gsea <- read.csv(no_psm_hallmark_file)
```

```r
no_psm_hallmark_gsea <- no_psm_hallmark_gsea %>%
  mutate(pathway_label = paste("No PSM Genes |",pathway_label))
```

```r
all_gsea <- rbind(hallmark_gsea, no_psm_hallmark_gsea)
```

## Format for output

```r
names(all_gsea)
```

```r
all_gsea <- all_gsea %>%
  rename(pathway = pathway_label) %>%
  select(-pathway_genes) %>%
  select(aifi_cell_type, timepoint, fg, bg,
         pathway, n_pathway_genes,
         NES, nomP, adjP,
         n_leadingEdge, leadingEdge)
```

```r
out_dir <- "output"
if(!dir.exists("output")) { dir.create("output") }
```

```r
out_file <- paste0("Supp-Table-06_Hallmark_GSEA_", Sys.Date(), ".csv")
write.csv(
    all_gsea,
    file.path(out_dir, out_file),
    row.names = FALSE,
    quote = FALSE
)
```

```r
system(paste('gzip', file.path(out_dir, out_file)))
```
